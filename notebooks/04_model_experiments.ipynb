```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import json

# Project root
BASE = Path("..")

# Dataset path
# DATA_PATH = BASE / "data" / "raw" / "patient_medicine_reviews.csv"
# DATA_PATH = BASE / "data" / "raw" / "patient_medicine_reviews_20000_synthetic.csv"
# DATA_PATH = BASE / "data" / "raw" / "patient_medicine_reviews_enhanced.csv"
# DATA_PATH = BASE / "data" / "raw" / "medireview_comprehensive_60000.csv"
DATA_PATH = BASE / "data" / "raw" / "pharmacy_review.csv"


# Load dataset
df = pd.read_csv(DATA_PATH)

df.head()
```

```python
import pandas as pd
from scipy.sparse import load_npz

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

# Load TF-IDF features
X_train = load_npz("../artifacts/X_train_tfidf.npz")
X_val = load_npz("../artifacts/X_val_tfidf.npz")

# Load labels
y_train = pd.read_csv("../data/processed/train.csv")["label"]
y_val = pd.read_csv("../data/processed/val.csv")["label"]

# Check shapes
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Create model
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# 2. Train model
lr_model.fit(X_train, y_train)

# 3. Predict validation data
y_pred = lr_model.predict(X_val)

# 4. Check accuracy
lr_accuracy = accuracy_score(y_val, y_pred)

print("Accuracy:", lr_accuracy)

# 5. Classification report
print("\nClassification Report:")
print(classification_report(y_val, y_pred, zero_division=0))

# 6. Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred))
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Create model
nb_model = MultinomialNB()

# 2. Train model
nb_model.fit(X_train, y_train)

# 3. Predict validation data
y_pred = nb_model.predict(X_val)

# 4. Calculate accuracy
nb_accuracy = accuracy_score(y_val, y_pred)

# 5. Display results
print("===== Naive Bayes =====")

print("\nAccuracy:", nb_accuracy)

print("\nClassification Report:")
print(classification_report(y_val, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred))
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Create model
svm_model = LinearSVC(
    max_iter=5000,
    random_state=42
)

# 2. Train model
svm_model.fit(X_train, y_train)

# 3. Predict validation data
svm_pred = svm_model.predict(X_val)

# 4. Calculate accuracy
svm_accuracy = accuracy_score(y_val, svm_pred)

# 5. Display results
print("===== Linear SVM =====")

print("\nAccuracy:", svm_accuracy)

print("\nClassification Report:")
print(classification_report(y_val, svm_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, svm_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Create model
dt_model = DecisionTreeClassifier(random_state=42)

# 2. Train model
dt_model.fit(X_train, y_train)

# 3. Predict validation data
y_pred = dt_model.predict(X_val)

# 4. Calculate accuracy
dt_accuracy = accuracy_score(y_val, y_pred)

# 5. Display results
print("===== Decision Tree =====")

print("\nAccuracy:", dt_accuracy)

print("\nClassification Report:")
print(classification_report(y_val, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Create model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# 2. Train model
rf_model.fit(X_train, y_train)

# 3. Predict validation data
y_pred = rf_model.predict(X_val)

# 4. Calculate accuracy
rf_accuracy = accuracy_score(y_val, y_pred)

# 5. Display results
print("===== Random Forest =====")

print("\nAccuracy:", rf_accuracy)

print("\nClassification Report:")
print(classification_report(y_val, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Create model
knn_model = KNeighborsClassifier(
    n_neighbors=5,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1
)

# 2. Train model
knn_model.fit(X_train, y_train)

# 3. Predict validation data
y_pred = knn_model.predict(X_val)

# 4. Calculate accuracy
knn_accuracy = accuracy_score(y_val, y_pred)

# 5. Display results
print("===== KNN =====")

print("\nAccuracy:", knn_accuracy)

print("\nClassification Report:")
print(classification_report(y_val, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred))
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score

# Get predictions from each trained model
nb_pred = nb_model.predict(X_val)
lr_pred = lr_model.predict(X_val)
dt_pred = dt_model.predict(X_val)
rf_pred = rf_model.predict(X_val)
knn_pred = knn_model.predict(X_val)

# SVM - only include this if you already trained svm_model
svm_pred = svm_model.predict(X_val)


# Create comparison table
results = pd.DataFrame({
    "Algorithm": [
        "Naive Bayes",
        "Logistic Regression",
        "Linear SVM",
        "Decision Tree",
        "Random Forest",
        "KNN"
    ],

    "Accuracy": [
        accuracy_score(y_val, nb_pred),
        accuracy_score(y_val, lr_pred),
        accuracy_score(y_val, svm_pred),
        accuracy_score(y_val, dt_pred),
        accuracy_score(y_val, rf_pred),
        accuracy_score(y_val, knn_pred)
    ],

    "Macro F1": [
        f1_score(y_val, nb_pred, average="macro"),
        f1_score(y_val, lr_pred, average="macro"),
        f1_score(y_val, svm_pred, average="macro"),
        f1_score(y_val, dt_pred, average="macro"),
        f1_score(y_val, rf_pred, average="macro"),
        f1_score(y_val, knn_pred, average="macro")
    ]
})

# Sort best Macro F1 first
results = results.sort_values("Macro F1", ascending=False)

print(results.round(4))
```

```python
import joblib

# Store all trained models
models = {
    "Naive Bayes": nb_model,
    "Logistic Regression": lr_model,
    "Linear SVM": svm_model,
    "Decision Tree": dt_model,
    "Random Forest": rf_model,
    "KNN": knn_model
}

# Create results table
results = pd.DataFrame({
    "Algorithm": [
        "Naive Bayes",
        "Logistic Regression",
        "Linear SVM",
        "Decision Tree",
        "Random Forest",
        "KNN"
    ],
    "Accuracy": [
        nb_accuracy,
        lr_accuracy,
        svm_accuracy,
        dt_accuracy,
        rf_accuracy,
        knn_accuracy
    ]
})

# Sort by accuracy (highest first)
results = results.sort_values(
    "Accuracy",
    ascending=False
)

print(results.round(4))

# Get best model name
best_name = results.iloc[0]["Algorithm"]

# Get best model
best_model = models[best_name]

# Save best model
joblib.dump(best_model, "../models/best_model.joblib")

# Save best model name
with open("../models/best_model_name.txt", "w") as file:
    file.write(best_name)

# Save model comparison results
results.to_csv(
    "../reports/validation_results.csv",
    index=False
)

print("\nBest Model:", best_name)
```
