```python
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import save_npz


# 1. Load cleaned data
df = pd.read_csv("../data/processed/cleaned_reviews.csv")


# 2. Split data: 60% train, 20% validation, 20% test
train, temp = train_test_split(
    df,
    test_size=0.4,
    random_state=42,
    stratify=df["label"]
)

val, test = train_test_split(
    temp,
    test_size=0.5,
    random_state=42,
    stratify=temp["label"]
)


# 3. Check sizes
print("Train:", len(train))
print("Validation:", len(val))
print("Test:", len(test))


# 4. Save splits
train.to_csv("../data/processed/train.csv", index=False)
val.to_csv("../data/processed/val.csv", index=False)
test.to_csv("../data/processed/test.csv", index=False)


# 5. Create TF-IDF
tfidf = TfidfVectorizer(
    max_features=3000,
    ngram_range=(1, 2),
    min_df=3,
    max_df=0.9
)


# 6. Fit TF-IDF ONLY on training data
X_train = tfidf.fit_transform(train["patient_review"])

# Use same TF-IDF on validation and test
X_val = tfidf.transform(val["patient_review"])
X_test = tfidf.transform(test["patient_review"])


# 7. Save TF-IDF features
save_npz("../artifacts/X_train_tfidf.npz", X_train)
save_npz("../artifacts/X_val_tfidf.npz", X_val)
save_npz("../artifacts/X_test_tfidf.npz", X_test)


# 8. Save TF-IDF model
joblib.dump(tfidf, "../artifacts/tfidf.joblib")


print("Training shape:", X_train.shape)
```
