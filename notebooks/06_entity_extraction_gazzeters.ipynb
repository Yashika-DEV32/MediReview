```python
medicines=['Albuterol',
    'Allopurinol',
    'Amlodipine',
    'Amoxicillin',
    'Aspirin',
    'Atorvastatin',
    'Azithromycin',
    'Budesonide',
    'Buspirone',
    'Cetirizine',
    'Clindamycin',
    'Colchicine',
    'Doxycycline',
    'Duloxetine',
    'Empagliflozin',
    'Escitalopram',
    'Fexofenadine',
    'Fluoxetine',]


dosage=[
    "10 mg","20 mg","30 mg","40 mg"
]

frequency=[
    "daily","weekly","monthly"
]


symptoms=[

    "vomit","dizziness","pain","nausea"
]



narrative="I have been prescribed Albuterol 20 mg for my nausea daily, and I also take Amlodipine for my blood pressure. Additionally, I use Cetirizine for my allergies and have been on Fluoxetine for my pain."


for doze in dosage:
    if doze in narrative:
        print(f"Dosage present in narrative :{doze}")

for freq in frequency:
    if freq in narrative:
        print(f"Frequency present in narrative :{freq}")

for symp in symptoms: 
    if symp in narrative:
        print(f"Symptom present in narrative :{symp}")



def narrative_analysis(narrative):
    mentioned_medicines = [medicine for medicine in medicines if medicine in narrative]
    mentioned_dosages = [doze for doze in dosage if doze in narrative]
    mentioned_frequencies = [freq for freq in frequency if freq in narrative]
    mentioned_symptoms = [symp for symp in symptoms if symp in narrative]

    return {
        "medicines": mentioned_medicines,
        "dosages": mentioned_dosages,
        "frequencies": mentioned_frequencies,
        "symptoms": mentioned_symptoms
    }

narrative="I have been prescribed Albuterol 20 mg for my nausea daily, and I also take Amlodipine for my blood pressure. Additionally, I use Cetirizine for my allergies and have been on Fluoxetine for my pain."
narrative_analysis_result = narrative_analysis(narrative)

print(f"Narrative Analysis Result: {narrative_analysis_result}")
```
